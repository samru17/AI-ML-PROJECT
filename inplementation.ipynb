```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
n = 1000
data = { "study_hours": np.random.randint(1, 10, n),
                  "attendance": np.random.randint(50, 101, n), 
                  "previous_marks": np.random.randint(30, 101, n),
                  "assignment_score": np.random.randint(30, 101, n),
                  "sleep_hours": np.random.randint(4, 10, n),
                  "internet_access": np.random.choice(["Yes", "No"], n),
                  "extracurricular": np.random.choice(["Yes", "No"], n) }
```

```python
df = pd.DataFrame(data)
```

```python
# Create result based on student performance
score = ( df["study_hours"] * 5 
          + df["attendance"] * 0.3
          + df["previous_marks"] * 0.4 
          + df["assignment_score"] * 0.2 )
```

```python
df["result"] = np.where(score >= 70, "Pass", "Fail")
df.to_csv("student_performance.csv", index=False)
```

```python
print("Dataset created successfully!")
print(df.head()) 
print("\nDataset shape:", df.shape)
print("\nResult count:")
print(df["result"].value_counts())
```
