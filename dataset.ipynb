```python
# Result count plot
ax = df["result"].value_counts().plot(kind="bar")
```

```python
ax.set_xlabel("Result")
ax.set_ylabel("Number of Students")
ax.set_title("Student Result Distribution")
for container in ax.containers:
    ax.bar_label(container)
    plt.show()
```
